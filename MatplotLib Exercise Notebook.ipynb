## Exercise 2 - Import Matplotlib using

import matplotlib.pyplot as plt



```python
import matplotlib.pyplot as plt
import pandas as pd

cars = pd.read_csv("car data.csv")
cars.head()
```

## Exercise 3 - Draw Barchart




```python
avg_price = cars.groupby("Fuel_Type")["Selling_Price"].mean().sort_values(ascending=False)
plt.figure(figsize=(7, 5))
plt.bar(avg_price.index, avg_price.values, color=["steelblue", "darkorange", "seagreen"])
plt.title("Average Selling Price by Fuel Type")
plt.xlabel("Fuel Type")
plt.ylabel("Average Selling Price")
plt.show()
```

## Exercise 4 -  Draw Histogram


```python
plt.figure(figsize=(8, 5))
plt.hist(cars["Selling_Price"], bins=20, color="seagreen", edgecolor="black")
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")
plt.show()
```

## Exercise 5 -  Draw Piechart



```python
transmission_counts = cars["Transmission"].value_counts()
plt.figure(figsize=(6, 6))
plt.pie(transmission_counts, labels=transmission_counts.index, autopct="%1.1f%%", startangle=90, colors=["steelblue", "darkorange"])
plt.title("Cars by Transmission Type")
plt.axis("equal")
plt.show()
```
