### Joining DataFrames (by index)

### Merging DataFrames (by a shared column)

```python
employees = pd.DataFrame({
    'EmployeeID': ['E001', 'E002', 'E003', 'E004', 'E005'],
    'Name': ['Kelly', 'Cat', 'Mat', 'Jhon', 'Sam'],
    'DeptID': ['D900', 'D800', 'D900', 'D700', 'D800']})

departments = pd.DataFrame({
    'DeptID': ['D900', 'D700', 'D800'],
    'DeptName': ['Computer', 'Accounts', 'HR']})

merged_df = pd.merge(employees, departments)
print(merged_df)
```

### In-Class Exercise 2: Clean Missing Data

```python
data = pd.DataFrame({
    "Student": ["Ana", "Ben", "Chris", "Divya", "Eli"],
    "Quiz1": [88, np.nan, 92, 81, 69],
    "Quiz2": [84, 80, np.nan, 79, 74]
})

print("Original rows:", len(data))

dropped = data.dropna()
print("Rows after dropna():", len(dropped))

filled = data.fillna(data[["Quiz1", "Quiz2"]].mean())
print("Rows after fillna():", len(filled))
print(filled)
```

**Discussion:** Dropping rows is better when only a few rows are incomplete, the data is plentiful, or a missing value makes the row unreliable (e.g. a missing target label). Filling is better when the dataset is small and every row matters, and a sensible replacement exists (mean/median for scores, 0 for "no purchases", the previous value for time series).

### Check Your Understanding
1. **Series vs DataFrame:** a Series is one-dimensional (a single labeled column); a DataFrame is two-dimensional with rows and columns (a table made of Series).
2. **Read a CSV:** `pd.read_csv()`.
3. **`df.iloc[:, 0:3]`:** all rows and the first three columns (positions 0, 1, 2).
4. **Handling missing values:** remove them with `dropna()` or replace them with `fillna()`.
5. **merge() vs slicing:** slicing only selects parts of one DataFrame; `merge()` is used when you need to bring in related data from another DataFrame using a shared key column (e.g. DeptID).
